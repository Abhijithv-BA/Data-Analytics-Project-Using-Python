**2. Data Pre-processing**

**2.1 Handling Missing Data**

Missing values are common in real-world datasets and can affect the accuracy of statistical analysis and machine learning models.

Therefore, the dataset is first examined to identify columns containing missing values. Since the dataset mainly consists of financial information, missing numerical values are handled using median imputation.

The median is selected instead of the mean because financial datasets can contain extreme values or outliers. The median is less sensitive to such extreme observations and therefore provides a more robust measure for replacing missing numerical values.

The missing values are replaced with the median value of their respective numerical columns.

**Check missing values**

```python
df.isnull().sum()
```

```python
# Identify numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

# Replace missing numerical values with the median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

Verify missing values after treatment

```python
df.isnull().sum()
```

**2.2 Handling Duplicate Records**

Duplicate records can lead to incorrect results by giving excessive weight to repeated observations. Therefore, duplicate records are identified and removed during the data cleaning process.

The duplicated() function is used to identify duplicate rows, while drop_duplicates() is used to remove them.

```python
df.duplicated().sum()
```

There are no duplicate rows present in the dataset. This dataset is clean in terms of duplicates and no further steps taken for duplicate removal.

**2.3 Data Formatting and Standardization**

Data formatting is performed to maintain consistency across the dataset. Text-based fields may contain unnecessary spaces or inconsistent formatting, which can create problems during grouping, filtering, and analysis.

The text columns are therefore cleaned by removing unnecessary leading and trailing spaces. Sector and industry names are also standardized to uppercase to maintain consistency.

Stock exchange codes are treated as identifiers rather than numerical variables because they are used to identify companies and are not intended for mathematical calculations.

**Identify text columns**

```python
text_columns = df.select_dtypes(include='object').columns
```

**Remove leading and trailing spaces**

```python
for column in text_columns:
    df[column] = df[column].str.strip()
```

**Standardize sector and industry**

```python
df['sector'] = df['sector'].str.upper()
df['industry'] = df['industry'].str.upper()
```

**Convert BSE code to text**

```python
df['BSE_code'] = df['BSE_code'].astype(str).str.strip()
```

**2.4 Creating Derived Columns**


**Feature Engineering**

Feature engineering involves creating new variables from existing data to provide additional analytical value.

Although the dataset already contains several financial indicators, additional features can be derived to support a deeper understanding of company performance.

The following derived features are created:

a. Net Profit Margin
measures the percentage of revenue converted into net profit.

b. Operating Expense Ratio
measures operating expenses as a percentage of revenue.

c. Profit Status
classifies companies as either Profit or Loss based on their net profit.

These derived variables can be useful during exploratory data analysis and visualization.

**a. Net Profit Margin**

```python
df['net_profit_margin'] = (
    df['net_profit'] / df['revenue']
) * 100
```

Explanation: Net Profit Margin indicates how much net profit a company generates from its revenue. A higher value generally indicates better profitability after accounting for expenses.

Formula:
Net Profit Margin = (Net Profit / Revenue) x 100

**b. Operating Expense Ratio**

```python
df['operating_expense_ratio'] = (
    df['operating_expenses'] / df['revenue']
) * 100
```

Explanation:
Operating Expense Ratio represents operating expenses as a percentage of revenue. It helps in understanding the proportion of revenue consumed by operating expenses.

Formula:
Operating Expense Ratio = (Operating Expenses / Revenue) x 100

**c. Profit Status**

```python
df['profit_status'] = np.where(
    df['net_profit'] > 0,
    'Profit',
    'Loss'
)
```

Explanation: The Profit Status feature categorizes companies based on their net profit. Companies with positive net profit are classified as "Profit", while companies with zero or negative net profit are classified as "Loss".

**3.13 Visualization 11 - Financial Correlation Heatmap**

This is particularly important for the multivariate analysis component.

```python
financial_columns = [
    'revenue',
    'operating_expenses',
    'operating_profit',
    'operating_profit_margin',
    'depreciation',
    'interest',
    'profit_before_tax',
    'tax',
    'net_profit',
    'EPS',
    'profit_TTM',
    'EPS_TTM'
]

correlation_matrix = df[financial_columns].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Financial Variables')
plt.show()
```

**Explanation:** Correlation Heatmap of Financial Variables. This is a multivariate analysis because it examines relationships among multiple numerical variables simultaneously. The rows and columns represent financial variables such as revenue, operating profit, net profit, interest and EPS. Each cell shows the correlation between two variables, with values ranging from minus one to plus one. Here used a heatmap because it provides a compact visual representation of many relationships at the same time. A positive correlation means the variables tend to move in the same direction, while a negative correlation means they tend to move in opposite directions.
